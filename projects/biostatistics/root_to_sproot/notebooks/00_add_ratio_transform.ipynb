- Just add the ratio of the sprout weight of total mass (transformation)  

    $Y = \frac{X_1}{X_1 + X_2}$        proportion of sprout mass in (total) biomass

- Where  

    $X_1 $ is sprout weight  
    $X_2 $ is root weight

```python
import pandas as pd
```

```python
path_data = "../data/dry_weight/pflanzen.tsv"
```

```python
df = pd.read_csv(path_data, sep="\t")
df.head()
```

```python
ratio = df["Spross"] / (df["Spross"] + df["Wurzel"])
df_ratio = pd.DataFrame(ratio, columns=["ratio"])
df = pd.concat([df, df_ratio], axis=1)
df.head()
```

```python
df.to_csv("../data/dry_weight/pflanzen_incl_ratio.csv",index=False)
```
