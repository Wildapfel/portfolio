```python
fig, axs = plt.subplots()
axs[0][0].pca_plot(df_GSE13159_train.iloc[:, 1:])
axs[0][0].pca_plot(df_GSE13159_test.iloc[:, 1:])
axs[0][0].pca_plot(df_GSE16455_test.iloc[:, 1:])
```

```python
fig, axs = plt.subplots()
labels =  ["GSE13159_train" for i in range(len(df_GSE13159_train))] 
labels += ["GSE13159_test" for i in range(len(df_GSE13159_test))]
labels += ["GSE16455_test" for i in range(len(df_GSE16455_test))]
labels = pd.DataFrame({"labels" : labels})
axs[0][0].pca_plot(pd.concat([df_GSE13159_train.iloc[:, 1:],
                              df_GSE13159_test.iloc[:, 1:], 
                              df_GSE16455_test.iloc[:, 1:]]),
                    labels = labels["labels"])
```
