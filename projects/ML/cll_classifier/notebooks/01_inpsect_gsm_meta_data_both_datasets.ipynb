```python
import GEOparse
```

```python
gsm = GEOparse.get_GEO(geo="GSM329417", destdir=None, how="full")
gsm.metadata
```

```python
if 'characteristics_ch1' in gsm.metadata:
    leukemia_class = gsm.metadata["characteristics_ch1"][-1].split(sep=":")[-1][1:]
    print(leukemia_class)
```
