```python
import pandas as pd
import plotwrapper.pyplot as plt
```

```python
df_GSE13159_train = pd.read_csv("/var/home/maxpetzold/HUB_local/projects/personal/transcriptomics/classifier/leukemia/v4.0/data/processed/bg_corrected_dqn_all_samples/GSE13159_train.csv")
df_GSE13159_test = pd.read_csv("/var/home/maxpetzold/HUB_local/projects/personal/transcriptomics/classifier/leukemia/v4.0/data/processed/bg_corrected_dqn_all_samples/GSE13159_test.csv")
df_GSE16455_test = pd.read_csv("/var/home/maxpetzold/HUB_local/projects/personal/transcriptomics/classifier/leukemia/v4.0/data/processed/bg_corrected_dqn_all_samples/GSE16455_test.csv")
df_GSE13159_train_y = pd.read_csv("/var/home/maxpetzold/HUB_local/projects/personal/transcriptomics/classifier/leukemia/v4.0/data/processed/bg_corrected_dqn_all_samples/GSE13159_train_y.csv")
df_GSE13159_test_y = pd.read_csv("/var/home/maxpetzold/HUB_local/projects/personal/transcriptomics/classifier/leukemia/v4.0/data/processed/bg_corrected_dqn_all_samples/GSE13159_test_y.csv")
df_GSE16455_test_y = pd.read_csv("/var/home/maxpetzold/HUB_local/projects/personal/transcriptomics/classifier/leukemia/v4.0/data/processed/bg_corrected_dqn_all_samples/GSE16455_test_y.csv")
```

```python
df_GSE13159_train_y.replace(0, "EXP01: CLL", inplace=True)
df_GSE13159_train_y.replace(1, "EXP01: Non-CLL", inplace=True)
df_GSE13159_test_y.replace(0, "EXP01: CLL", inplace=True)
df_GSE13159_test_y.replace(1, "EXP01: Non-CLL", inplace=True)
df_GSE16455_test_y.replace(0, "EXP02: CLL", inplace=True)
df_GSE16455_test_y.replace(1, "EXP02: Non-CLL", inplace=True)
```

```python
df_cat = pd.concat([df_GSE13159_train.iloc[:, 1:], df_GSE13159_test.iloc[:, 1:], df_GSE16455_test.iloc[:, 1:]])
df_cat_labels = pd.concat([df_GSE13159_train_y, df_GSE13159_test_y, df_GSE16455_test_y])
```
